```python
import os
from sklearn.model_selection import train_test_split
import random
import numpy as np
import torch
from torch.utils.data import random_split

def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

set_seed(42)

# Define original paths
image_dir = '/kaggle/input/cloud-masking-dataset/content/train/data'
mask_dir = '/kaggle/input/cloud-masking-dataset/content/train/masks'

# Get image–mask pairs
paired_files = [f for f in image_files if os.path.exists(os.path.join(mask_dir, f))]

# Split data
train_size = int(0.8 * len(paired_files))
test_size = len(paired_files) - train_size

train_files, test_files = random_split(
    paired_files,
    [train_size, test_size],
    generator=torch.Generator().manual_seed(42)  # <--- fixed seed
)
# Let's say you want 80% train and 20% val from the training set
val_ratio = 0.2
val_size = int(val_ratio * len(train_files))
train_size = len(train_files) - val_size

train_subfiles, val_subfiles = random_split(
    train_files,
    [train_size, val_size],
    generator=torch.Generator().manual_seed(42)  # fixed seed
)

# Create full paths for training and validation
train_image_paths = [os.path.join(image_dir, f) for f in train_subfiles]
train_mask_paths = [os.path.join(mask_dir, f) for f in train_subfiles]

val_image_paths = [os.path.join(image_dir, f) for f in val_subfiles]
val_mask_paths = [os.path.join(mask_dir, f) for f in val_subfiles]

# train_files, test_files = train_test_split(paired_files, test_size=0.2, random_state=42)

test_image_paths = [os.path.join(image_dir, f) for f in test_files]
test_mask_paths = [os.path.join(mask_dir, f) for f in test_files]

print(f"Train images: {len(train_image_paths)}, Val images: {len(val_image_paths)}, len(Test images: {len(test_image_paths)}")
```

### Data Loading

```python
import os
import torch
import numpy as np
from osgeo import gdal
from torch.utils.data import Dataset, DataLoader
class CloudSegmentationDataset(Dataset):
    def __init__(self, image_paths, mask_paths):
        self.image_paths = image_paths
        self.mask_paths = mask_paths

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        # Load image
        img_ds = gdal.Open(self.image_paths[idx], gdal.GA_ReadOnly)
        img_array = img_ds.ReadAsArray()  # Shape: (bands, H, W)
        # Ensure image has at least 4 bands
        if img_array.shape[0] < 4:
            raise ValueError(f"Image at {self.image_paths[idx]} has less than 4 bands")

        # Select bands 1 and 4 → (2, H, W)
        selected_bands = np.stack([img_array[0], img_array[3]], axis=0)
        
        image_tensor = torch.tensor(selected_bands, dtype=torch.float32)

        # Load mask
        mask_ds = gdal.Open(self.mask_paths[idx], gdal.GA_ReadOnly)
        mask_array = mask_ds.ReadAsArray()  # Shape: (H, W)
        mask_tensor = torch.tensor(mask_array, dtype=torch.long)

        return image_tensor, mask_tensor, self.image_paths[idx], self.mask_paths[idx]
```

```python
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
import joblib
from tqdm import tqdm
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
import joblib
from tqdm import tqdm
import gc  # Garbage collector
def dice_coefficient(mask1: np.ndarray, mask2: np.ndarray) -> float:
    """Computes the Dice coefficient between two binary masks."""
    intersection = np.sum(mask1 * mask2)
    return (2.0 * intersection + 1e-7) / (np.sum(mask1) + np.sum(mask2) + 1e-7)  # Avoid division by zero

class MemorySafeCloudSegmenter:
    def __init__(self):
        self.model = None
        self.scaler = StandardScaler()
        
    def _extract_features(self, band1, band4):
        """Memory-safe feature extraction with NDVI"""
        ndvi = (band4 - band1) / (band4 + band1 + 1e-10)
        return np.column_stack([band1.flatten(), band4.flatten(), ndvi.flatten()])
    
    def train_in_batches(self, dataset, batch_size, num_estimators,max_dpth, max_smples):
        """
        Train model using batched processing to save memory
        
        Args:
            dataset: Your CloudSegmentationDataset
            batch_size: Number of images to process at once
            subsample: Fraction of pixels to use from each image
        """
        # Initialize model with reduced parameters
        self.model = RandomForestClassifier(
            n_estimators=num_estimators,  # Reduced from 100
            max_depth=max_dpth,       # Reduced from 10
            max_samples=max_smples,   # Use only 50% of samples per tree
            n_jobs=-1,
            random_state=42
        )
        
        # Process in batches
        for batch_start in tqdm(range(0, len(dataset), batch_size)):
            batch_features = []
            batch_labels = []
            
            # Process each image in batch
            for i in range(batch_start, min(batch_start+batch_size, len(dataset))):
                image, mask, _, _ = dataset[i]
                band1 = image[0].numpy()
                band4 = image[1].numpy()
                mask = mask.numpy().flatten()
                
                # Extract features
                features = self._extract_features(band1, band4)
                
                batch_features.append(features)
                batch_labels.append(mask)
                
                # Explicit cleanup
                del band1, band4, features
                gc.collect()
            
            # Scale and partial fit
            batch_features = np.concatenate(batch_features)
            batch_labels = np.concatenate(batch_labels)
            
            # Scale features in place to save memory
            if batch_start == 0:
                self.scaler.fit(batch_features)
            
            batch_features = self.scaler.transform(batch_features)
            
            # Train on batch
            self.model.fit(batch_features, batch_labels)
            
            # Clean up
            del batch_features, batch_labels
            gc.collect()
    
    def predict_image_safely(self, image_tensor, chunk_size=50000):
        """Memory-safe prediction for single image"""
        band1 = image_tensor[0].numpy()
        band4 = image_tensor[1].numpy()
        h, w = band1.shape
        
        # Process in chunks
        features = self._extract_features(band1, band4)
        features = self.scaler.transform(features)
        
        pred = np.zeros(features.shape[0], dtype=np.uint8)
        for i in range(0, len(features), chunk_size):
            chunk = features[i:i+chunk_size]
            pred[i:i+chunk_size] = self.model.predict(chunk)
            del chunk
            gc.collect()
        
        return pred.reshape((h, w))
    
    def evaluate_with_dice(self, dataset, max_images=None):
        """Memory-safe evaluation with Dice coefficient"""
        dice_scores = []
        eval_range = range(min(len(dataset), max_images) if max_images else len(dataset))
        
        for i in tqdm(eval_range):
            image, mask, _, _ = dataset[i]
            pred = self.predict_image_safely(image)
            true = mask.numpy()
            
            dice = dice_coefficient(true, pred)
            dice_scores.append(dice)
            
            # Clean up
            del image, mask, pred, true
            gc.collect()
        
        return np.mean(dice_scores)
```

```python
# Initialize with reduced memory footprint
segmenter = MemorySafeCloudSegmenter()

# Train with small batches and heavy subsampling
segmenter.train_in_batches(
    train_dataset,
    batch_size=2,      # Process 2 images at a time
    num_estimators=50,
    max_dpth=8,
    max_smples=0.5
)

# Evaluate on subset if needed
val_dice = segmenter.evaluate_with_dice(val_dataset)
print(f"Validation Dice: {val_dice:.4f}")

# Save memory by deleting training data if no longer needed
del train_dataset
gc.collect()

# Full evaluation when needed
test_dice = segmenter.evaluate_with_dice(test_dataset)
print(f"Test Dice: {test_dice:.4f}")

# Save model
joblib.dump(segmenter, 'memory_safe_segmenter.pkl')
# from itertools import product
# import gc
# from tqdm import tqdm
# import joblib

# # Define your parameter grid
# param_grid = {
#     'n_estimators': [30, 50],  # Fewer options to save memory
#     'max_depth': [5, 8],
#     'max_samples': [0.3, 0.5]
# }

# # Generate all parameter combinations
# param_names = param_grid.keys()
# param_values = product(*param_grid.values())

# best_dice = 0.0
# best_model = None
# best_params = None

# for values in tqdm(list(param_values), desc="Hyperparameter search"):
    # params = dict(zip(param_names, values))
    
    # # Clear memory before each training run
    # gc.collect()
    
    # # Initialize segmenter
    # segmenter = MemorySafeCloudSegmenter()
    
    # # Train with current parameters
    # segmenter.train_in_batches(
    #     train_dataset,
#         batch_size=2,      # Process 2 images at a time
#         num_estimators=params['n_estimators'],
#         max_dpth=params['max_depth'],
#         max_smples=params['max_samples']
#     )
    
#     # Evaluate on validation set
#     val_dice = segmenter.evaluate_with_dice(val_dataset, max_images=10)
#     print(f"Params {params} -> Validation Dice: {val_dice:.4f}")
    
#     # Track best model
#     if val_dice > best_dice:
#         best_dice = val_dice
#         best_model = segmenter.model
#         best_params = params
#         print(f"New best! Dice: {val_dice:.4f}")
        
#     # Clean up
#     del segmenter
#     gc.collect()
        
# # After tuning, train final model with best parameters if needed
# if best_model is not None:
#     print(f"\nBest parameters found: {best_params}")
#     print(f"Best validation Dice: {best_dice:.4f}")
    
#     # Initialize final segmenter with best model
#     final_segmenter = MemorySafeCloudSegmenter()
#     final_segmenter.model = best_model
    
#     # Evaluate on test set
#     test_dice = final_segmenter.evaluate_with_dice(test_dataset)
#     print(f"Test Dice: {test_dice:.4f}")
    
#     # Save model
#     joblib.dump(final_segmenter.model, 'best_memory_safe_segmenter.pkl')
# else:
#     print("No successful model training completed")
```
